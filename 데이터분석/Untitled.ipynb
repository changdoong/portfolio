### csv 구조 확인

```python
import pandas as pd
import os

files = {
    "countries": "countries.csv",
    "education_technology": "education_technology.csv",
    "environment_energy": "environment_energy.csv",
    "gdp_economy": "gdp_economy.csv",
    "population_health": "population_health.csv"
}

dfs = {}

for name, filename in files.items():

    file_path = os.path.join(path, filename)

    dfs[name] = pd.read_csv(file_path)

    print(f"\n{'='*50}")
    print(f"파일명: {filename}")
    print("크기:", dfs[name].shape)
    print("컬럼:", dfs[name].columns.tolist())

    display(dfs[name].head(3))
```

### 분석용 데이터셋 만들기

```python
# 1. 분석 대상 국가
target_countries = ["KOR", "USA", "JPN", "DEU"]
```

```python
# 2. 국가 정보 필터링
country_info = dfs["countries"]
country_info = country_info[country_info["country_id"].isin(target_countries)].copy()

display(country_info)
```

```python
# 3. 연도별 데이터 병합
merge_keys = ["country_id", "year"]

data = dfs["gdp_economy"].copy()

for name in [
    "education_technology",
    "environment_energy",
    "population_health"
]:
    data = data.merge(
        dfs[name],
        on=merge_keys,
        how="left"
    )
```

```python
# 4. 국가 정보 병합
data = data.merge(
    country_info[
        ["country_id", "country_name", "region", "income_level"]
    ],
    on="country_id",
    how="left"
)
```

```python
# 5. 기간 및 국가 필터링
data = data[
    (data["country_id"].isin(target_countries)) &
    (data["year"].between(2015, 2023))
].copy()

# 6. 정렬
data = data.sort_values(
    ["country_id", "year"]
).reset_index(drop=True)

print("데이터 크기:", data.shape)
display(data.head(10))
```

## 데이터 품질 검사

```python
# 국가별 연도 개수
print("국가별 관측치")
display(
    data.groupby("country_name")["year"].nunique()
)

# 국가별 연도 범위
display(
    data.groupby("country_name")["year"].agg(["min", "max", "count"])
)

# 결측치 비율
missing = (
    data.isnull()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
    .round(2)
)

display(
    missing.to_frame("missing_pct")
)

# 중복 데이터 확인
print("중복 행:", data.duplicated(["country_id", "year"]).sum())
```

```python
# 국가별 연도 누락 확인

expected_years = set(range(2015, 2024))

for country in data["country_name"].unique():
    actual_years = set(
        data.loc[data["country_name"] == country, "year"]
    )

    missing_years = sorted(expected_years - actual_years)

    print(f"{country}")
    print(f"누락 연도: {missing_years}")
    print("-" * 30)
```
