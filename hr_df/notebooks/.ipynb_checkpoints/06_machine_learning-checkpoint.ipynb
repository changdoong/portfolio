## 1. 분석 목적 및 문제 정의

### 분석 목적

본 단계에서는 직원의 인구통계학적 특성, 직무 특성, 급여 및 근무 환경 데이터를 활용하여 직원의 이직 여부를 예측하는 분류 모델을 구축한다.

앞선 EDA와 통계 분석에서 확인한 변수 간 관계를 바탕으로 머신러닝 모델을 학습하고, 다양한 알고리즘의 예측 성능을 비교한다.

### 문제 정의

- 문제 유형: 이진 분류(Binary Classification)
- 예측 대상: Attrition
- 클래스 1: 이직(Yes)
- 클래스 0: 재직(No)

### 주요 분석 목표

1. 직원의 이직 여부를 예측할 수 있는 모델 구축
2. 여러 분류 알고리즘의 성능 비교
3. 클래스 불균형을 고려한 모델 평가
4. 이직자 탐지 성능을 높일 수 있는 모델 탐색

### 분석 시 유의사항

본 데이터는 공개된 HR 분석용 샘플 데이터이므로 실제 기업 전체의 인력 특성을 대표한다고 볼 수 없다.

또한 모델의 예측 결과는 실제 이직 원인을 입증하는 것이 아니며, 관측된 변수에 기반한 예측 결과로 해석한다.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

# 데이터 분할
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate

# 전처리
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# 모델
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier

# 평가 지표
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)

# 시각화 설정
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.unicode_minus"] = False
sns.set_theme(style="whitegrid")
```

## 3. 데이터 로드

02_data_preprocessing.ipynb에서 전처리한 데이터를 불러온다.

이번 단계에서는 기존 전처리 데이터를 기반으로 머신러닝 학습에 적합한 입력 변수(X)와 예측 대상(y)을 구성한다.

```python
hr_df = pd.read_csv("../data/processed/hr_preprocessed.csv")

print("데이터 크기:", hr_df.shape)

display(hr_df.head())
```

## 4. 타깃 변수 및 입력 변수 확인

예측 대상은 직원의 이직 여부(Attrition)로 설정한다.

이직 여부를 제외한 변수 중 직원 식별번호와 같이 예측에 활용할 수 없는 변수는 제거한다.

특히 EmployeeNumber는 직원 식별을 위한 고유 번호이므로 모델 학습 변수에서 제외한다.

```python
# 타깃 변수
target = "Attrition"

# 모델 학습에서 제외할 변수
drop_cols = [
    "Attrition",
    "AttritionBinary",
    "EmployeeNumber"
]

# 입력 변수와 타깃 변수 분리
X = hr_df.drop(columns=drop_cols)
y = hr_df[target].map({"Yes": 1, "No": 0})

print("입력 변수 크기:", X.shape)
print("타깃 변수 크기:", y.shape)

# 타깃 변수와 동일한 정보가 입력 변수에 남아 있는지 확인
print("AttritionBinary 포함 여부:", "AttritionBinary" in X.columns)

display(X.head())
```

## 5. 클래스 불균형 확인

직원 이직 데이터는 일반적으로 재직자보다 이직자의 비율이 낮을 수 있다.

이러한 클래스 불균형은 모델이 다수 클래스인 재직자를 중심으로 학습하게 만들 수 있다.

따라서 모델 학습에 앞서 타깃 변수의 클래스별 데이터 수와 비율을 확인한다.

```python
class_count = y.value_counts().sort_index()

class_ratio = y.value_counts(normalize=True).sort_index() * 100

class_result = pd.DataFrame({
    "Employee Count": class_count,
    "Ratio (%)": class_ratio
})

class_result.index = ["No Attrition", "Attrition"]

display(
    class_result.style
    .background_gradient(cmap="YlGnBu")
    .format({"Ratio (%)": "{:.2f}%"})
)
```

```python
sns.countplot(x=y, hue=y, legend=False)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition (0: No, 1: Yes)")
plt.ylabel("Employee Count")
plt.show()
```

## 6. Train / Test Split

모델의 일반화 성능을 평가하기 위해 데이터를 학습 데이터와 테스트 데이터로 분리한다.

- 학습 데이터: 80%
- 테스트 데이터: 20%
- random_state: 2020158011

또한 이직 여부의 클래스 비율이 학습 데이터와 테스트 데이터에 최대한 동일하게 유지되도록 Stratified Sampling을 적용한다.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=2020158011,
    stratify=y
)

print("학습 데이터:", X_train.shape)
print("테스트 데이터:", X_test.shape)

print("\n학습 데이터 이직률:", round(y_train.mean() * 100, 2), "%")
print("테스트 데이터 이직률:", round(y_test.mean() * 100, 2), "%")
```

## 7. 변수 유형 분류 및 전처리

머신러닝 모델은 문자열 형태의 범주형 데이터를 직접 학습할 수 없으므로 수치형 데이터로 변환하는 과정이 필요하다.

본 분석에서는 변수의 특성에 따라 다음과 같이 전처리한다.

- 범주형 변수: One-Hot Encoding
- 수치형 변수: StandardScaler

전처리 과정은 ColumnTransformer와 Pipeline을 활용하여 구성한다.

이를 통해 학습 데이터에서만 전처리 기준을 학습하고, 테스트 데이터에는 동일한 기준을 적용하여 데이터 누수를 방지한다.

```python
# 수치형 변수
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# 범주형 변수
categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("수치형 변수 수:", len(numeric_features))
print("범주형 변수 수:", len(categorical_features))

print("\n수치형 변수")
print(numeric_features)

print("\n범주형 변수")
print(categorical_features)
```

```python
# 수치형 변수 전처리
numeric_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)

# 범주형 변수 전처리
categorical_transformer = Pipeline(
    steps=[
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

# 전체 전처리
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("전처리 파이프라인")
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("학습 데이터 전처리 결과:", X_train_processed.shape)
print("테스트 데이터 전처리 결과:", X_test_processed.shape)
```

## 8. 전처리 결과 확인

전처리 이후 데이터의 차원과 변환 결과를 확인한다.

One-Hot Encoding을 적용하면 범주형 변수의 범주별로 새로운 컬럼이 생성되므로 기존 데이터보다 컬럼 수가 증가할 수 있다.

또한 수치형 변수는 StandardScaler를 통해 평균 0, 표준편차 1을 기준으로 변환된다.

```python
# 전처리 후 컬럼명 확인
feature_names = preprocessor.get_feature_names_out()

print("전처리 후 전체 변수 수:", len(feature_names))

display(
    pd.DataFrame({
        "Feature": feature_names
    }).head(20)
)
```

## 9. Logistic Regression Baseline Model

Logistic Regression은 이진 분류 문제에서 널리 사용되는 선형 분류 알고리즘이다.

본 분석에서는 Logistic Regression을 기준 모델(Baseline Model)로 설정하여 직원의 이직 여부를 예측한다.

### 모델 구축 목적

- 기본적인 이직 예측 성능 확인
- 이후 앙상블 모델과의 성능 비교 기준 마련
- 선형 모델을 통한 분류 가능성 확인

클래스 불균형에 따른 영향을 확인하기 위해 우선 기본 설정으로 모델을 학습하고, 이후 평가 지표를 통해 성능을 분석한다.

```python
lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_model.fit(X_train_processed, y_train)
```

```python
lr_pred = lr_model.predict(X_test_processed)

# 이직 확률 예측
lr_proba = lr_model.predict_proba(X_test_processed)[:, 1]

display(
    pd.DataFrame({
        "Actual": y_test.values,
        "Predicted": lr_pred,
        "Attrition Probability": lr_proba
    }).head(10)
)
```

## 10. Baseline Model Evaluation

학습한 Logistic Regression 모델의 예측 성능을 평가한다.

본 분석에서는 다음 지표를 활용한다.

- Accuracy: 전체 예측 중 정답 비율
- Precision: 이직으로 예측한 직원 중 실제 이직자의 비율
- Recall: 실제 이직자 중 모델이 찾아낸 비율
- F1-score: Precision과 Recall의 조화평균
- ROC-AUC: 다양한 분류 임계값에서 모델의 구분 능력을 평가

특히 이직자를 놓치지 않는 것이 중요하므로 Recall과 F1-score를 함께 중점적으로 확인한다.

```python
lr_results = {
    "Model": "Logistic Regression",
    "Accuracy": accuracy_score(y_test, lr_pred),
    "Precision": precision_score(y_test, lr_pred, zero_division=0),
    "Recall": recall_score(y_test, lr_pred, zero_division=0),
    "F1-score": f1_score(y_test, lr_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_test, lr_proba)
}

lr_result_df = pd.DataFrame([lr_results])

display(
    lr_result_df.style
    .background_gradient(cmap="YlGnBu", subset=[
        "Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"
    ])
    .format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)
```

```python
print(
    classification_report(
        y_test,
        lr_pred,
        target_names=["No Attrition", "Attrition"],
        zero_division=0
    )
)
```

## 11. Confusion Matrix

Confusion Matrix를 통해 실제 이직 여부와 모델의 예측 결과를 비교한다.

특히 실제 이직자를 재직자로 잘못 예측한 False Negative(FN)의 규모를 확인한다.

이는 이직 가능성이 있는 직원을 탐지하지 못한 사례를 의미한다.

```python
cm = confusion_matrix(y_test, lr_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Attrition", "Attrition"]
)

disp.plot(cmap="Blues", values_format="d")

plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

그러나 실제 이직자를 탐지하는 Recall은 42.55%로 상대적으로 낮게 나타났다.

이는 모델이 전체적인 분류 성능은 확보했지만, 실제 이직자를 탐지하는 데에는 한계가 있음을 보여준다.

따라서 이후 모델링에서는 클래스 불균형 대응과 분류 임계값 조정을 고려하여 이직자 탐지 성능을 개선하고자 한다.

본 모델은 이후 앙상블 모델과의 성능 비교를 위한 Baseline으로 활용한다.

## 12. Optuna를 활용한 하이퍼파라미터 최적화

기본 머신러닝 모델의 성능을 개선하기 위해 Optuna를 활용한 하이퍼파라미터 최적화를 진행한다.

Optuna는 탐색 과정에서 이전 실험 결과를 반영하여 다음 하이퍼파라미터 조합을 탐색하는 최적화 라이브러리이다.

본 분석에서는 Random Forest의 주요 하이퍼파라미터를 탐색하고, Stratified K-Fold 교차검증을 통해 모델의 일반화 성능을 평가한다.

최적화 목표는 F1-score로 설정한다.

```python
import optuna
from sklearn.model_selection import cross_val_score
```

## 13. Random Forest 하이퍼파라미터 최적화

Random Forest의 성능에 영향을 미치는 주요 하이퍼파라미터를 탐색한다.

### 탐색 대상

- n_estimators
- max_depth
- min_samples_split
- min_samples_leaf
- max_features
- class_weight

### 최적화 방법

Stratified K-Fold 교차검증을 활용하여 각 파라미터 조합의 F1-score 평균을 계산한다.

또한 전처리 과정에서 발생할 수 있는 데이터 누수를 방지하기 위해 전처리 파이프라인과 모델을 하나의 Pipeline으로 구성한다.

```python
# 교차검증 설정
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

def objective(trial):

    params = {
        "n_estimators": trial.suggest_int(
            "n_estimators", 100, 500
        ),

        "max_depth": trial.suggest_int(
            "max_depth", 3, 20
        ),

        "min_samples_split": trial.suggest_int(
            "min_samples_split", 2, 20
        ),

        "min_samples_leaf": trial.suggest_int(
            "min_samples_leaf", 1, 10
        ),

        "max_features": trial.suggest_categorical(
            "max_features", ["sqrt", "log2"]
        ),

        "class_weight": trial.suggest_categorical(
            "class_weight", [None, "balanced", "balanced_subsample"]
        ),

        "random_state": 42,
        "n_jobs": 1
    }

    # 전처리 + 모델 파이프라인
    model = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(**params))
    ])

    # 교차검증
    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="f1",
        n_jobs=-1
    )

    return scores.mean()
```

## 14. Optuna 최적화 실행

Optuna의 TPE 샘플러를 사용하여 하이퍼파라미터를 탐색한다.

총 30회의 실험을 진행하고, 각 실험에서 계산된 교차검증 F1-score를 비교하여 최적의 파라미터 조합을 선정한다.

```python
study = optuna.create_study(
    direction="maximize",
    sampler=optuna.samplers.TPESampler(seed=42)
)

study.optimize(
    objective,
    n_trials=30
)

print("최적 F1-score:", study.best_value)
print("최적 파라미터:")

for key, value in study.best_params.items():
    print(f"{key}: {value}")
```

```python
# 최적화 과정 시각화
optuna.visualization.matplotlib.plot_optimization_history(study)

plt.title("Optuna Optimization History")
plt.show()
```

```python
# 파라미터 중요도
optuna.visualization.matplotlib.plot_param_importances(study)

plt.show()
```

```python
best_params = study.best_params.copy()

best_rf = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        **best_params,
        random_state=42,
        n_jobs=-1
    ))
])

best_rf.fit(X_train, y_train)

# 테스트 데이터 예측
rf_optuna_pred = best_rf.predict(X_test)

rf_optuna_proba = best_rf.predict_proba(X_test)[:, 1]
```

```python
rf_optuna_results = {
    "Model": "Random Forest + Optuna",
    "Accuracy": accuracy_score(y_test, rf_optuna_pred),
    "Precision": precision_score(y_test, rf_optuna_pred, zero_division=0),
    "Recall": recall_score(y_test, rf_optuna_pred, zero_division=0),
    "F1-score": f1_score(y_test, rf_optuna_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_test, rf_optuna_proba)
}

rf_optuna_result_df = pd.DataFrame([rf_optuna_results])

display(
    rf_optuna_result_df.style
    .background_gradient(
        cmap="YlGnBu",
        subset=["Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"]
    )
    .format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)
```

## 15. 모델 성능 비교 및 최종 평가

Logistic Regression 기준 모델과 Optuna를 활용하여 하이퍼파라미터를 최적화한 Random Forest 모델의 성능을 비교한다.

동일한 테스트 데이터를 활용하여 Accuracy, Precision, Recall, F1-score, ROC-AUC를 비교하고 각 모델의 예측 성능을 평가한다.

특히 직원 이직 예측이라는 분석 목적을 고려하여 Recall과 F1-score를 중요하게 살펴본다.

```python
# 기존 모델과 Optuna 모델 결과 통합
comparison_df = pd.concat(
    [lr_result_df, rf_optuna_result_df],
    ignore_index=True
)

# 성능 비교 테이블
display(
    comparison_df.style
    .background_gradient(
        cmap="YlGnBu",
        subset=["Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"]
    )
    .format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)
```

### 15-1. 모델별 성능 시각화

두 모델의 평가 지표를 시각화하여 각 모델의 강점과 약점을 비교한다.

```python
metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-score",
    "ROC-AUC"
]

comparison_plot = comparison_df.set_index("Model")[metrics]

ax = comparison_plot.T.plot(
    kind="bar",
    figsize=(12, 6),
    rot=0
)

plt.title("Model Performance Comparison")
plt.ylabel("Score")
plt.xlabel("Evaluation Metrics")
plt.ylim(0, 1.05)
plt.legend(title="Model")
plt.grid(axis="y", alpha=0.3)
plt.show()
```
